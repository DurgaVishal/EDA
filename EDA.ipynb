```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
df =pd.read_csv("/content/titanic.csv")
df.head() #used to display first 5 rows
df.info() #used to get full details
df
print(df.shape)
df.info()
df.describe()
df.isnull().sum()
df["Embarked"]=df["Embarked"].fillna(
    df["Embarked"].mode()[0]
)
print(df.isnull().sum())
print("Before:",df.shape[0])  #o - checks the complete rows number, 1- gives column number or count( values before removing duplicate )

df=df.drop_duplicates()
print("After:",df.shape[0])
df=df.rename(columns={
   "Passengerid":"PassengerID",
   "2urvived":"Survived"
})

df["Sex"]= df["Sex"].map({
    0:"male",
    1:"female"

})
df.head()
sns.histplot(data=df,x="Age", bins=20)

plt.title("Age Distribution of Titanic Passengers")
plt.xlabel("Age")
plt.ylabel("Number of Passengers")
plt.show()
sns.countplot(data=df,x="Sex",hue="Survived")

plt.title("Titanic Survival by Passenger Class")
plt.xlabel("Sex")
plt.ylabel("number of passengers")
plt.show()

sns.countplot(data=df,x="Pclass", hue="Survived")
plt.title("titanic survival by passenger class")
plt.xlabel("Passeneger class")
plt.ylabel("number of passengers")
plt.show()
```
